```python
# Define the module for performing log-rank tests on survival data extracted from MMP Pickle files.

import pickle
import numpy as np
from lifelines.statistics import logrank_test
from lifelines import KaplanMeierFitter
import matplotlib.pyplot as plt

def read_test_results_pkl(file_path):
    """
    Load a Pickle file containing prediction results.
    """
    print(f"Loading file: {file_path}")
    try:
        with open(file_path, 'rb') as f:
            data = pickle.load(f)
        return data
    except Exception as e:
        print(f"Error reading Pickle file: {e}")
        return None

def extract_survival_data_from_mmp_pkl(pkl_file_path):
    """
    Extract survival data needed for log-rank test from MMP Pickle results.
    """
    pkl_data = read_test_results_pkl(pkl_file_path)
    if pkl_data is None:
        return None

    test_data = pkl_data["test"]
    try:
        risk_scores = test_data["all_risk_scores"].flatten()
        event_times = test_data["all_event_times"].flatten()
        event_indicators = (1 - test_data["all_censorships"]).flatten()

        return {
            "risk_scores": risk_scores,
            "event_times": event_times,
            "event_indicators": event_indicators,
            "sample_ids": test_data.get("sample_ids", []),
        }
    except Exception as e:
        print(f"Error processing data: {e}")
        return None


# Log-rank test function
def perform_mmp_logrank_test_minimal(survival_data):
    risk_scores = survival_data["risk_scores"]
    event_times = survival_data["event_times"]
    event_indicators = survival_data["event_indicators"]

    threshold = np.percentile(risk_scores, 50)
    high_mask = risk_scores >= threshold
    low_mask = risk_scores < threshold

    high_times = event_times[high_mask]
    high_events = event_indicators[high_mask]
    low_times = event_times[low_mask]
    low_events = event_indicators[low_mask]

    results = logrank_test(
        high_times, low_times,
        event_observed_A=high_events,
        event_observed_B=low_events
    )

    print(f"Log-rank p-value: {results.p_value:.2e}")
    print(f"Significant: {'Yes' if results.p_value < 0.05 else 'No'}")

    return results
```

```python
# BRCA & Transformer-based MMP results
fold_files = [
    "../logs/dumps/brca_trans/k=0/all_dumps.h5",
    "../logs/dumps/brca_trans/k=1/all_dumps.h5",
    "../logs/dumps/brca_trans/k=2/all_dumps.h5",
    "../logs/dumps/brca_trans/k=3/all_dumps.h5",
    "../logs/dumps/brca_trans/k=4/all_dumps.h5",
]

# Prepare lists to collect data
all_risk_scores = []
all_event_times = []
all_event_indicators = []

# Loop through each fold
for file_path in fold_files:
    data = extract_survival_data_from_mmp_pkl(file_path)
    if data is None:
        continue
        
    all_risk_scores.append(data["risk_scores"])
    all_event_times.append(data["event_times"])
    all_event_indicators.append(data["event_indicators"])

# Concatenate arrays
import numpy as np

risk_scores = np.concatenate(all_risk_scores)
event_times = np.concatenate(all_event_times)
event_indicators = np.concatenate(all_event_indicators)

# Build survival data dictionary
survival_data = {
    "risk_scores": risk_scores,
    "event_times": event_times,
    "event_indicators": event_indicators,
}
# Run test
results = perform_mmp_logrank_test_minimal(survival_data)
```
