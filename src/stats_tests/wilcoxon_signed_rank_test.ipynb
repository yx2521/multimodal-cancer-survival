```python
from scipy.stats import wilcoxon
```

```python
# BRCA Wilcoxon test 1 VS 2 modalities

one_modality_brca_scores = [0.56701, 0.64154, 0.67652, 0.7489, 0.65651]
two_modality_brca_scores = [0.7543, 0.79232, 0.8248, 0.79736, 0.70526]
stat, p = wilcoxon(two_modality_brca_scores, one_modality_brca_scores, alternative='greater')
print(f"BRCA 1 VS 2 Wilcoxon p-value: {p:.4f}")
```
