```python
import pandas as pd
import os
import numpy as np

# Download pancancer-normalized expression data from Xena database for TCGA BRCA cohort
!mkdir -p ../data_csvs/rna/hallmarks/BRCA

# Download TCGA-BRCA transcriptomic expression data (pancancer-normalized)
!curl -o ../data_csvs/HiSeqV2_PANCAN.BRCA.gz https://tcga-xena-hub.s3.us-east-1.amazonaws.com/download/TCGA.BRCA.sampleMap%2FHiSeqV2_PANCAN.gz
!gunzip ../data_csvs/HiSeqV2_PANCAN.BRCA.gz

# Download Hallmark gene signatures
!curl -L -o ../data_csvs/rna/metadata/hallmarks_signatures.csv https://raw.githubusercontent.com/mahmoodlab/SurvPath/main/datasets_csv/metadata/hallmarks_signatures.csv
```

```python
# read pan-cancer normalized RNA data 
df = pd.read_csv('../data_csvs/HiSeqV2_PANCAN.BRCA', delimiter='\t')

# read 
hallmarks = pd.read_csv('../data_csvs/rna/metadata/hallmarks_signatures.csv').values.flatten()
hallmarks = hallmarks[~pd.isnull(hallmarks)]
hallmarks = pd.unique(hallmarks)

# filter out and transpose
df = df[df['sample'].isin(hallmarks)].reset_index(drop=True)
df = df.set_index('sample')
df = df.transpose()
df = df.sort_index(axis=1)
df = df.reset_index()
df = df.rename(columns={'index': 'sample'})
```

```python
df.to_csv('../data_csvs/rna/hallmarks/BRCA/rna_clean.csv')
df
```
